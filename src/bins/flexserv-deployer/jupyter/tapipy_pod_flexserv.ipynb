```python
from getpass import getpass
from tapipy.tapis import Tapis

tenant = 'dev.develop'

base_host = f'{tenant}.tapis.io'

base_url = f'https://{base_host}'

# Enter Tapis Username. Example: trainingXX
tapis_username = input('Username: ')
# Enter Tapis password. Example: trainingXX
tapis_password = getpass(prompt='Password: ', stream=None)

t = Tapis(
    base_url=base_url,
    username=tapis_username,
    password=tapis_password,
    grant_type="password"
)

t.get_tokens()

jwt = t.access_token.access_token

jwt
```

```python


model_info = {
    'model_id' : "Qwen/Qwen3-0.6B", 
    'sha' : "c1899de289a04d12100db370d81485cdf75e47ca"
}

short_sha = model_info['sha'][:8]

volume_spec = {
    'volume_id': short_sha,
    'description': 'volume for storing models',
    'size_limit': 10240 # size of zip file + size of unzipped model + 1024 MB buffer
}

pod_spec = {
    'pod_id': short_sha,
    'image': 'zhangwei217245/ai-serving-py-transformers:latest',
    'command': ['python', '/app/main.py'],
    'arguments': [],
    'environment_variables': {
        'MODEL_HASH': model_info['sha'],
        'MODEL_DIR': '/models',
    },
    'volume_mounts': {
        f'{volume_spec["volume_id"]}': {
            "type": "tapisvolume",
            "mount_path": "/models"
        }
    },
    'networking': {
        "default": {
            "protocol": "http",
            "port": 8000
        }
    },
    'resources': {
        "cpu_request": 10001,
        "cpu_limit": 10001,
        "mem_request": 15001,
        "mem_limit": 15001,
        "gpus": 0
    },
    'description': 'Pod for AI inference serving'
}

```

```python
model_local_path = f'{Path.home()}/.models/{model_info["sha"]}.zip'
model_zip_filename = model_info['sha'] + '.zip'

# Upload the model file (temporary, not persistent)
with open(model_local_path, 'rb') as file:
    t.pods.upload_to_volume(
        volume_id=volume_spec['volume_id'],
        path=model_zip_filename,
        file=file
    )


files = t.pods.list_volume_files(volume_id=volume_spec['volume_id'], path='/')
print(files)
```
